```python
#2
import numpy as np
expression = np.array([2.5, 7.1, 0.8, 5.4, 9.2, 3.0, 6.5, 4.1], dtype = np.float64)
expression_copy = np.array([2.5, 7.1, 0.8, 5.4, 9.2, 3.0, 6.5, 4.1], dtype = np.float32)
print(expression.nbytes)
print(expression_copy.nbytes)
expressionx2 = expression * 2
print(expressionx2)
sr = sum(expression)/len(expression)
new_expression = expression - sr
print(new_expression)
print(min(expression))
print(max(expression))
print(sr)

expression_p = [2.5, 7.1, 0.8, 5.4, 9.2, 3.0, 6.5, 4.1]
sr_p = sum(expression_p)/len(expression_p)
expx2 = []
for x in expression_p:
    expx2.append(x * 2)
print(expx2)
new_exp = []
for x in expression_p:
    new_exp.append(x - sr_p)
print(new_exp)
```

```python
#3
import numpy as np
milk = np.array([
    [20, 22, 21],
    [25, 24, 26],
    [18, 19, 20],
    [30, 28, 29]
])
correction = np.array([1, 2, -1, 0])
print(milk.shape)
print(correction.shape)
new_milk = milk + correction.reshape(4, 1)
print(new_milk)
sr1 = np.mean(new_milk[0])
sr2 = np.mean(new_milk[1])
sr3 = np.mean(new_milk[2])
sr4 = np.mean(new_milk[3])
print(sr1, sr2, sr3, sr4)
day1 = np.mean(new_milk[:, 0])
day2 = np.mean(new_milk[:, 1])
day3 = np.mean(new_milk[:, 2])
print(day1, day2, day3)
summ1 = np.sum(new_milk[0])
summ2 = np.sum(new_milk[1])
summ3 = np.sum(new_milk[2])
summ4 = np.sum(new_milk[3])
print(summ1, summ2, summ3, summ4)
summ = [summ1, summ2, summ3, summ4]
print(1 + np.argmax(summ))
```
